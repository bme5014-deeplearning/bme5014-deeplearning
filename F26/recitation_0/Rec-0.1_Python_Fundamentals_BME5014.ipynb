## 2. Variables, Types & Indentation

Python is **dynamically typed** so you don't declare a type, it's inferred at runtime. The four basic types you'll use constantly: `int`, `float`, `str`, `bool`.

Python also uses **indentation**, not curly braces, to define code blocks. Indentation is not a style choice here, it's syntax. Get it wrong and your code either errors out or silently does the wrong thing.


```python
heart_rate_bpm = 78       # int
body_temp_c = 36.6        # float
patient_id = "P-2044"     # str
is_critical = False       # bool

print(type(heart_rate_bpm), type(body_temp_c), type(patient_id), type(is_critical))

# Indentation defines the block:
if is_critical:
    print("ALERT: notify attending physician")
else:
    print("Vitals stable")
```

## 3. Operators

Standard arithmetic, plus two Python-specific ones worth knowing cold: **floor division** (`//`) and **exponentiation** (`**`). You'll use both constantly once you're computing output shapes for the CNN layers you'll build to process MRI or ultrasound images.


```python
samples, batch_size = 10, 3

print("samples + batch_size  =", samples + batch_size)
print("samples - batch_size  =", samples - batch_size)
print("samples * batch_size  =", samples * batch_size)
print("samples / batch_size  =", samples / batch_size)    # true division -> float
print("samples // batch_size =", samples // batch_size)   # floor division -> int (rounds down)
print("samples % batch_size  =", samples % batch_size)    # modulo (remainder)
print("samples ** 2          =", samples ** 2)             # exponent
```

## 4. Booleans & Logical Operators

`and`, `or`, `not` are used everywhere in conditionals, loops, and filtering logic — including building the inclusion/exclusion criteria you'd use to filter a patient cohort.


```python
has_fever = True
has_cough = False

print("has_fever and has_cough:", has_fever and has_cough)  # False - both must be True
print("has_fever or has_cough :", has_fever or has_cough)    # True  - only one needs to be True
print("not has_fever          :", not has_fever)             # False - flips the value
```

## 5. Strings & f-strings

Strings are **immutable**. Methods like `.lower()`, `.strip()`, `.split()` return a *new* string rather than modifying in place. `.strip()` in particular will save you constantly when cleaning messy clinical data exports (stray whitespace in CSV columns is extremely common).

**f-strings** (`f"..."`) let you embed variables and expressions directly inside a string, including format specifiers like decimal precision — handy for printing lab values without twenty decimal places.


```python
s = "  Patient Report, Case 001  "
print(s.lower())
print(s.strip())
print(s.strip().split(","))

patient_name = "Ahmet"
age = 45
print(f"Patient {patient_name}, age {age}.")

import math
print(f"Pi to 2 decimals: {math.pi:.2f}")
print(f"Pi to 5 decimals: {math.pi:.5f}")
```

## 6. Data Structures: Lists, Tuples, Dicts, Sets

| Structure | Ordered | Mutable | Duplicates | Syntax |
|---|---|---|---|---|
| `list` | Yes | Yes | Yes | `[1, 2, 3]` |
| `tuple` | Yes | **No** | Yes | `(1, 2, 3)` |
| `dict` | Yes (3.7+) | Yes | Unique keys | `{"a": 1}` |
| `set` | No | Yes | **No** | `{1, 2, 3}` |


```python
# Lists are mutable and ordered - a running log of heart-rate readings
vitals_log = [72, 75, 78]
vitals_log.append(80)
print("vitals_log:", vitals_log)

# Tuples are immutable and ordered. Good for fixed records like a date of birth.
patient_dob = (1990, 5, 12)  # (year, month, day)
try:
    patient_dob[0] = 1991
except TypeError as e:
    print("Tuple error:", e)

# Dictionaries are key/value pairs - a single patient record
patient = {"patient_id": "P-2044", "diagnosis": "Type 2 Diabetes"}
print("dict:", patient, "| diagnosis:", patient["diagnosis"])

# Sets are unique, unordered - de-duplicating diagnosis codes across visits
icd_codes = {"E11", "I10", "E11", "J45"}
print("set:", icd_codes)
```

## 7. Comprehensions

A compact way to build a list/dict/set from an iterable in a single line. You'll see these constantly in homework starter code for cleaning and transforming clinical data, so it's worth being comfortable reading and writing them.


```python
temps_f = [98.6, 101.2, 99.5, 103.1, 97.8]  # patient temperatures in Fahrenheit

temps_c = [(f - 32) * 5 / 9 for f in temps_f]
fevers_f = [f for f in temps_f if f > 100.4]              # > 38C is considered a fever
patient_to_temp_c = {f"P{i+1}": round((f - 32) * 5 / 9, 1) for i, f in enumerate(temps_f)}

print("temps_c:          ", temps_c)
print("fevers_f:         ", fevers_f)
print("patient_to_temp_c:", patient_to_temp_c)
```

## 8. Conditionals

`if` / `elif` / `else`. Only the first `True` branch runs; the rest are skipped. A very common real use: classifying a reading into clinical categories.


```python
systolic_bp = 145  # mmHg

if systolic_bp < 120:
    print("Normal")
elif systolic_bp < 130:
    print("Elevated")
elif systolic_bp < 140:
    print("Hypertension Stage 1")
else:
    print("Hypertension Stage 2")
```

## 9. Loops

Use a `for` loop when you know what you're iterating over; use `while` when you're looping until some condition changes. `range(5)` starts at `0` and stops *before* `5`, that's the single most common off-by-one mistake in this class, so get used to it now.

`enumerate` and `zip` are two built-ins you'll reach for a lot: `enumerate` gives you the index alongside the value, `zip` lets you iterate over two sequences in parallel (e.g. patient IDs alongside their readings).

**Heads up:** in deep learning, your data is often huge — a single ECG recording can have tens of thousands of samples, and an MRI volume can have millions of voxels. Looping in pure Python over every element is slow. Whenever you can, you'll want **vectorized** NumPy/PyTorch operations instead of explicit loops (more on this in the NumPy section below).


```python
for day in range(5):
    print(day, end=" ")
print()  # 0 1 2 3 4 --> never reaches 5

x = 3
while x > 0:
    print(x, end=" ")
    x -= 1
print()

patient_ids = ["P-001", "P-002", "P-003"]
for i, pid in enumerate(patient_ids):
    print(i, pid)

heart_rates = [90, 85, 70]
for pid, hr in zip(patient_ids, heart_rates):
    print(pid, "->", hr, "bpm")
```

## 10. Functions, *args / **kwargs, and Type Hints

`def` defines a function. `*args` collects extra positional arguments into a tuple; `**kwargs` collects extra keyword arguments into a dict. This is what lets a function accept a variable number of inputs — very handy when a patient record has a variable number of symptoms or vitals attached.

Type hints (`def f(x: int) -> int:`) are optional and **not enforced at runtime**, they're purely for readability and for your IDE/type checker. Get used to reading them; most homework starter code uses them.


```python
def greet(patient_name):
    print(f"Hello {patient_name}")

greet("Bob")

def describe_patient(patient_id, *symptoms, **vitals):
    print("ID:", patient_id)
    print("symptoms:", symptoms)
    print("vitals:", vitals)

describe_patient("P-2044", "cough", "fatigue", heart_rate=92, temp_c=38.1)

def add(a: int, b: int) -> int:
    return a + b

print(add(2, 3))
```

## 11. Mutability & Pass-by-Reference

Variables in Python are passed **by reference**. For mutable objects like lists, a function can modify the original object in place (changes persist outside the function). This trips people up constantly when debugging — for example when a function is supposed to just *read* a monitoring log but accidentally mutates it — so internalize it now.


```python
def log_reading(readings):
    readings.append(100)

vitals_log = [72, 75, 78]
log_reading(vitals_log)
print(vitals_log)  # [72, 75, 78, 100] --> modified in place
```

## 12. Classes: a First Look

A class bundles data (attributes) and behavior (methods) together. `__init__` is the constructor, called automatically when you create an instance. `self` refers to the specific instance.

This is intentionally a light touch — inheritance, encapsulation, abstraction, and polymorphism are deeper OOP concepts you'll pick up as needed later in the course.


```python
class Patient:
    def __init__(self, patient_id, heart_rate):
        self.patient_id = patient_id
        self.heart_rate = heart_rate

    def summary(self):
        print(f"Patient {self.patient_id}: heart rate {self.heart_rate} bpm.")

p = Patient("P-2044", 92)
p.summary()
```

## 13. Error Handling

`try` / `except` / `finally` lets you handle errors gracefully instead of crashing. `finally` always runs, whether or not an error occurred. Errors you'll see often this semester: `FileNotFoundError`, `KeyError`, `ValueError`, `RuntimeError` (very common with PyTorch shape mismatches), and `TypeError`.


```python
def dose_per_kg(total_mg, weight_kg):
    try:
        return total_mg / weight_kg
    except ZeroDivisionError:
        print("Missing/invalid weight. Returning None")
        return None
    finally:
        print("dose_per_kg finished")

print(dose_per_kg(500, 70))
print(dose_per_kg(500, 0))
```

## 16. NumPy Basics

This is the section to actually get comfortable with. NumPy arrays are the mental model for PyTorch tensors, and you'll be reasoning about shapes constantly — whether that's a 1D ECG signal, a 2D image slice, or a 3D imaging volume.


```python
import numpy as np

# Creating arrays & indexing - five consecutive resting heart-rate readings
hr = np.array([72, 74, 76, 78, 80])
print("hr    :", hr)
print("third :", hr[2])
print("last  :", hr[-1])
```

```python
# Shape and reshaping - think of this as a tiny 3x3 patch of pixel intensities
# from a grayscale medical image (e.g. one small patch of an MRI slice)
patch = np.arange(9).reshape(3, 3)
print(patch)
print("shape:", patch.shape)
print("center pixel patch[1,1]:", patch[1, 1])
```

```python
# Slicing - extracting a row of pixels, a column of pixels, and a small
# receptive-field-like block, all from the same image patch
M = np.arange(1, 10).reshape(3, 3)
print(M)

row1 = M[0, :]
col1 = M[:, 0]
block = M[:2, :2]

print("row1 :", row1)
print("col1 :", col1)
print("block:\n", block)
```

```python
# Elementwise operations & broadcasting - e.g. combining two channels/readings
x = np.array([1, 2, 3])
y = np.array([10, 20, 30])

print("x + y :", x + y)
print("x ** 2:", x ** 2)
```

```python
# Elementwise (*) vs matrix multiplication (@)
# Hadamard product: exactly how you'd apply a binary segmentation MASK to an
# image, zeroing out background pixels while keeping the region of interest.
# Matrix multiplication: this is literally the computation inside every
# linear/fully-connected layer you will implement later in this course
# (output = weights @ input).
A = np.array([[1, 2], [3, 4]])
B = np.array([[2, 0], [1, 2]])

hadamard = A * B   # elementwise product (e.g. image * mask)
matmul = A @ B      # matrix multiplication (e.g. weights @ input)

print("Hadamard (A * B):\n", hadamard)
print("Matmul   (A @ B):\n", matmul)
```
